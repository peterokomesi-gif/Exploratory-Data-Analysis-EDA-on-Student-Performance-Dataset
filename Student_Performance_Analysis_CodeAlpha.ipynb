**Count Students Who Scored 0**

```python
failed_students = (df['overall_score'] == 0).sum()

print("Number of students who scored 0:", failed_students)
```

**Visualization for Failed Students**

```python
labels = ['Failed (0)', 'Passed (>0)']

values = [
    (df['overall_score'] == 0).sum(),
    (df['overall_score'] > 0).sum()
]

plt.figure(figsize=(6,5))

sns.barplot(x=labels, y=values)

plt.title("Failed vs Passed Students")
plt.ylabel("Number of Students")

plt.show()
```

Bar Chart **Comparison**

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7,5))

sns.barplot(x='internet_access',
            y='overall_score',
            data=df)

plt.title("Internet Access vs Final Grade (Overall Score)")
plt.xlabel("Internet Access at Home")
plt.ylabel("Average Final Grade")

plt.show()
```

```python
df['internet_access'].value_counts()
```

```python
from scipy.stats import ttest_ind

internet_yes = df[df['internet_access'] == 'Yes']['overall_score']
internet_no = df[df['internet_access'] == 'No']['overall_score']

t_stat, p_value = ttest_ind(internet_yes, internet_no)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```
